```python
import sqlite3
import pandas as pd

# Kết nối đến file database của bạn
conn = sqlite3.connect('database/amazon_reviews.db')

# 1. Thống kê số lượng review theo từng mức star_rating (1 đến 5 sao)
query_rating = """
    SELECT star_rating, COUNT(*) AS total_reviews 
    FROM amazon_reviews 
    GROUP BY star_rating 
    ORDER BY star_rating DESC;
"""
df_rating = pd.read_sql(query_rating, conn)
print("Thống kê số lượng theo star_rating:")
print(df_rating)

conn.close()
```
